# 1. Syntaxe

Il n'y a pas de séparateur d'instructions. On écrit une instruction par ligne. Il est possible de répartir une instruction sur plusieurs lignes avec la barre oblique inversée `\` (*backslash*).

Les commentaires sont matérialisés par le caractère dièse `#` (*sharp*). Aucun caractère à droite du caractère # n'est interprété. 
Pour faire des commentaires multi-lignes, on encadre le commentaire ainsi <code>'''commentaire'''</code>.

Le nom d'une variable peut utiliser les caractères alphanumériques : `A->Z`, `a->z`, `0->9`) et le tiret bas `_` (*underscore*). Les variables doivent commencer par une lettre ou un tiret bas mais pas par un chiffre (on fait en général débuter les noms de variables par une minuscule ou un tiret bas ; il est possible mais pas recommandé d'utiliser des lettres accentuées). Python est sensible à la casse : les variables `abc`, `ABC`, `Abc` sont distinctes.

On peut affecter un objet à une variable avec le signe égal `=` ; il est possible d'effectuer des affectations multiples sur une seule ligne en les séparant par une virgule `,`.

Il existe en Python une trentaine de mots-clefs réservés (instructions ou fonctions) qui ne doivent pas être utilisés comme noms de variable.

Par exemple,
- `print()` est une fonction (en Python 3, une instruction en Python 2) qui permet d'afficher la valeur d'un objet ;
- `type()` est une fonction qui permet d'afficher le type ou la classe d'un objet ;
- `dir()` est une fonction qui permet d'énumérer les propriétés (attributs ou méthodes) d'un objet ;
- `help()` est une fonction qui permet d'afficher une aide sur un objet ;

````python
# premières lignes de code Python
a = 1  # on affecte à la variable a la valeur 1
b, c = 2, 3  # affectations multiples
print(a, b, c)# affichage de la valeur des variables a, b et c
````

````python
type(a)  # de quel type est la variable a ?

# NB: pour accéder aux méthodes (seulement aux méthodes) d'un objet, 
# on peut aussi utiliser la touche tabulation après avoir écrit 'nom_objet.'
````

````python
dir(a)  # énumération des attributs et méthode de l'entier a
````

````python
help(a)  # aide sur l'entier a
````

````python
# aide en ligne du notebook
a?
````

## 2.1 Absence de valeur

`None` est la seule instance du type `NoneType`. Cet objet permet de matérialiser l'absence de valeur.

On utilise cet objet lorsque l'on a besoin d'une variable sans nécessairement lui affecter de valeur initiale.

````python
a = None
print(a)
````

## 2.2 Booléens

Le type `bool` a deux instances :

- `True` : valeur logique vrai
- `False` : valeur logique faux

Les opérateurs logiques sont :

- `and` : ET logique, *a and b*
- `or` : OU logique, *a or b*
- `not` : NON logique, *non a*

````python
p = True
q = False
print(p and q, p or q, not p, not q)
````

<div class="alert alert-info">
    <h3><i class="fa fa-info-circle"></i>  Astuce</h3>
    <ul>
        <li>Quelques objets sont considérés comme <code>False</code> comme les zéros et les collections vides.</li>
        <li>Tous les autres objets sont considérés comme <code>True</code>.</li>
        <li>Les booléens sont implémentés comme des entiers: <code>True</code> vaut <code>1</code> et <code>False</code> vaut <code>0</code>.</li>
        <li>L'interprète Python effectue une évaluation paresseuse (<i>lazzy evaluation</i>) du second argument des opérations logiques <code>and</code> et <code>or</code> :
            <ul>
                <li>p <code>and</code> q : si p est <code>False</code>, q n'est pas évalué</li>
                <li>p <code>or</code> q : si p est <code>True</code>, q n'est pas évalué</li>
            </ul>
    </ul>
</div>

### *list*

Un objet de type `list` est une séquence ordonnée *modifiable*. La séquence vide s'écrit `[]`.

**Usage** : Une liste permet de stocker des valeurs, par exemple les noms des colonnes d'un fichier CSV :

```Python
cols = ["Date", "Open", "High", "Low", "Close", "Adj Close", "Volume"]
```

````python
liste1 = [1, 2, 3]
print(liste1)
````

#### Quelques méthodes utiles

````python
liste1.append(4)  # ajout d'un élément
print(liste1)
````

````python
liste1.extend([5, 6, 7])  # extension avec une autre liste
print(liste1)
````

#### Accesseur aux sous-parties d'une séquence (liste, tuple ou chaîne)

L'opérateur `[]` permet d'accéder à différentes sous-parties d'une liste en fonction des indices de ses éléments (les indices démarrent à 0) :
- `a[i]` : accès au i ème élément
- `a[i:j]` : accès à la sous-liste allant du ième élément au j-1ème élément inclus
- `a[i:j:k]` : accès à la sous-liste allant du ième élément au j-1ème élément inclus par pas de k

Par convention :
- `a[i:]` vaut `a[i:len(a)]`
- `a[:j]` vaut `a[0:j]`

````python
# utilisation de l'opérateur []

liste2 = ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"]

print(liste2[2])
print(liste2[1:3])
print(liste2[1:4:2])
print(liste2[1:])
print(liste2[:3])
````

#### Autres méthodes utiles

- `insert(i, val)` insertion d'une valeur à un indice
- `pop(i)` extraction de la valeur à la position
- `reverse()` retournement de la liste

### *range*

Un objet de type `range` est une séquence ordonnée *immuable* qui représente une liste d'entiers compris entre 2 entiers (débutant à 0 par défaut) avec un pas (de 1 par défaut).

La séquence s'arrête juste avant la valeur maximale.

**Usage** : Cet objet est généralement utilisé avec une boucle `for` (cf. `02_Python.ipynb`).

```Python
for i in range(10):
    print(i)
```

````python
range1 = range(10)  # les entiers de 0 à 9
range2 = range(1, 11)  # les entiers de 1 à 10
range3 = range(0, 100, 2)  # les entiers pairs de 0 à 100 exclu
print(list(range1))
print(list(range2))
print(list(range3))
````

### *str*

Un objet de type `str` représente une chaîne de caractères (par défaut encodée en UTF-8 en Python 3). Une chaîne vide s'écrit `''` ou `""`.

Il y a plusieurs manières d'écrire les chaînes en Python :

- avec une apostrophe `'` ou un guillement `"`
- avec trois apostrophes `'''` ou trois guillemets `"""` pour les chaînes s'étendant sur plusieurs lignes

Attention, pour utiliser un caractère spécial dans une chaîne (apostrophe, guillement, backslash) il faut placer le caractère backslash avant le caractère.

Echappements dans les chaînes de caractères :
- `\\` : backslash
- `\'` : apostrophe
- `\"` : guillement
- `\b` : espace
- `\n` : saut de ligne (LF)
- `\r` : retour charriot (CR)
- `\t` : tabulation (HT)

Attention aux chemins Windows qui utilisent le backslash : `C:\\Users\\Guest\\`

````python
chaine1 = 'ceci est une chaîne de caractères'
chaine2 = "c'est bien une chaîne"  # pas besoin de backslash car le délimiteur est un guillement
chaine3 = 'c\'est aussi une chaîne'  # backslash indispensable car le délimiteur est une apostrophe
chaine4 = 'C:\\Users\\Guest\\'  # chemin Windows
chaine5 = '''lundi,
mardi,
mercredi,
...'''  # chaîne sur plusieurs lignes
print(chaine1, chaine2, chaine3, chaine4, chaine5, sep='\n')
````

### *dict*

Un objet de type `dict` représente une liste associative de type clef&rarr;valeur. Un objet de type `dict` n'est pas ordonné. Le dictionnaire vide s'écrit `{}`.

On accès à une valeur en utilisant l'opérateur `[]`. Il est possible de supprimer une valeur avec l'opérateur `del`.

**Usage** : Un dictionnaire peut être utilisé pour établir des correspondances entre des valeurs :

```Python
iso_3166 = {"DE": "Allemagne",
            "FR": "France",
            "IT": "Italie",
            ...
           }
```

````python
dico1 = {'un': 1, 'deux': 2, 'trois': 3}  # litéral avec des accolades
dico2 = dict(un=1, deux=2, trois=3)  # dict() avec des mots-clefs
dico3 = dict([('deux', 2), ('un', 1), ('trois', 3)])  # liste de 2-tuples
dico4 = dict(zip(['un', 'deux', 'trois'], [1, 2, 3]))  # fonction zip
dico5 = {'trois': 3, 'un': 1, 'deux': 2}  # ordre indifférent
print(dico1, dico2, dico3, dico4, dico5)
````

````python
dico1 == dico2 == dico3 == dico4 == dico5
````

````python
# accesseur
dico1["un"]
````

````python
# accesseur avec valeur par défaut
dico1.get("un", 0)
````

````python
del dico1["un"]  # suppression d'une clef
dico1
````

````python
# accesseur avec valeur par défaut
dico1.get("un", 0)
````

````python
# mise à jour
dico1["quatre"] = 4
dico1.update({"deux": 2.0, "six": 6}) # ou bien dico1 = dico1 | {"deux": 2.0, "six": 6}
dico1
````

<div class="alert alert-info">
    <h3><i class="fa fa-info-circle"></i>  Astuce</h3>
    <ul>
        <li>L'opérateur <code>|</code> permet de créer un nouveau dictionnaire à partir de 2 dictionnaires en priorisant les clés du second :</li>
        <code>{"a": 1, "b": 2} | {"b": 3, "c": 4}</code> => <code>{"a": 1, "b": 3, "c": 4}</code>
    </ul>
</div>

````python
d1 ={"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}
print(d1|d2, d1, d2)
````

````python
d1 |= d2
print(d1, d2)
````

**Usage** : Un set peut être utilisé pour stocker les identifiants des enregistrements d'un fichier :

````python
ensemble = set()
ensemble.add(1)
ensemble.add(2)
ensemble.add(2)
ensemble.update([1, 3, 4])
ensemble
````
