# 0. Rappels, prérequis ou compétences à acquérir

## Système & terminal
- Arborescence fichiers et de répertoires, chemins absolus/relatifs
- Commandes de base (linux, bash) : `pwd`, `cd`, `ls`, `touch`, `chmod`, `rm`, `mkdir`, `rmdir`, `mv`, `cat`, `grep`, `find`
- Variables d'environnement : `PATH`
- Différences Windows / macOS / Linux : encodages, séparateurs
- Auto-formation (linux) :
    - Console linux : <https://pixees.fr/informatiquelycee/n_site/nsi_prem_cmd_base_linux.html>
    - datacamp/Shell : <https://app.datacamp.com/learn/courses/introduction-to-shell>
- Alternatives :
    - Windows : cmd, PowerShell

## Encodages & données
- Chaînes de caractères : ASCII, Unicode, UTF-8
- Nombres à virgule flottante (norme IEEE 754) et précision limitée (`0.1 + 0.2`)
- Dates, heures et fuseaux horaires, UTC, GMT (norme ISO 8601)
- Pays (norme ISO 3166)
- Latitudes et longitudes (normes WGS84, ISO 6709)

## Web & formats de données
- HTML/CSS notions de base
- Couleurs : RVB, hexadécimal, HSV
- Formats de données : JSON, XML, YAML, CSV
- Protocoles : HTTP/HTTPS, codes de statut de réponse (200, 404, 500), URL
- Notions d'API REST : endpoint, requête GET/POST, headers

## Algorithmique de base
- Structures de données classiques : tableau, pile (LIFO), file (FIFO), table de hachage
- Complexité : $O(1)$, $O(log(n))$, $O(\sqrt n)$, $O(n)$, $O(n.log(n))$, $O(n^2)$, $O(2^n)$, $O(n!)$
- Récursivité
- Cryptographie : fonction de hachage, symétrique, asymétique (clé publique, clé privée)

## Cloud computing
- IaaS, PaaS, SaaS
- GAFAM : Amazon Web Services, Google Cloud, Microsoft Azure
- Souverains : OVHcloud, Scaleway, etc.
- Pour Python : Colab, Pythonanywhere, etc.
