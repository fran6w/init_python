## 1.1 `if`

L'instruction `if` suivie d'un booléen (ou de n'importe quel objet) permet d'exécuter du code de manière conditionnelle.

Elle peut être complétée par une ou plusieurs instructions `elif` suivie(s) d'un booléen et également par une instruction `else` pour les cas où aucune condition précédente n'est remplie. Toutes ces instructions nécessitent le caractère `:` en fin de ligne. 

Le code conditionné doit ensuite être indenté (tabulation ou nombre identiques d'espaces) par rapport au code principal. Il est possible d'imbriquer des instructions `if` en indentant correctement à chaque fois.

````python
age = 17

if age in range(10, 20):
    print("teenager")
````

````python
if age < 18:
    print("mineur")
else:
    print("majeur")
````

````python
if age < 10:
    print("enfant")
elif age < 20:
    print("teenager")
else:
    print("adulte")
````

````python
# idem avec des if imbriqués
if age < 20:
    if age < 10:
        print("enfant")
    else:
        print("teenager")
else:
    print("adulte")
````

````python
"mineur" if age < 18 else "majeur"  # sur une seule ligne
````

## 1.2 `for`

L'instruction `for` permet de boucler sur du code à partir d'un objet de type séquence. Elle s'utilise avec l'instruction `in` et nécessite le caractère `:` en fin de ligne. La séquence de code en boucle doit être indentée. Si l'interprèteur rencontre l'instruction `break`, la boucle est interrompue. Si l'interprèteur rencontre l'instruction `continue` il passe directement à l'itération suivante.


Il est possible de compléter l'instruction `for` avec une instruction `else`. Le code correspondant au `else` sera exécutée après le `for` sauf si ce dernier a été interrompu par un `break`.

&#9888; Attention de ne pas modifier une séquence sur laquelle on effectue une boucle.

````python
for a in [1, 2, 3]:
    print(a * a)
````

````python
for a in range(1, 10, 2):
    print(a * a)
````

````python
for i in range(6):
    if i == 2:
        continue
    if i > 4:
        break
    print(i)
````

La fonction `enumerate()` permet de boucler à la fois sur l'indice (commençant à 0 par défaut) et les valeurs d'une séquence.

````python
# boucle avec indice
for i, a in enumerate("Python"):  # indices + éléments ; le tuple (i, a) est implicite
    print(i, a)
````

La fonction `zip()` permet de boucler sur plusieurs séquences jusqu'à épuisement de la séquence la plus courte.

````python
# boucle sur plusieurs séquences
for x, y, z in zip(range(10), list("Python 3.13"), range(20, 10, -1)):
    print(x, y, z)
````

#### Compréhension de liste

Il s'agit d'un moyen de construire des listes de manière très concise. Une application classique est la construction de nouvelles listes où chaque élément est le résultat d'une opération appliquée à chaque élément d'une autre séquence, ou bien de créer une sous-séquence des éléments satisfaisant une condition spécifique.

````python
# exemple de compréhension de liste
[i * i for i in range(10)]
````

## 1.3 `while`

L'instruction `while` suivi d'un booléen (ou de n'importe quel objet) permet de boucler sur une séquence de code tant que le booléen a pour valeur `True`. Elle nécessite le caractère `:` en fin de ligne. La séquence de code répété doit être indentée. Si l'interprèteur rencontre l'instruction `break`, la boucle est interrompue. Si l'interprèteur rencontre l'instruction `continue` il passe directement à l'itération suivante.

Attention, si la condition est toujours vraie, la boucle ne s'arrêtera jamais sauf avec un `*break`.

Il est possible de compléter l'instruction `while` avec une instruction `else`. Le code correspondant au `else` sera exécuté après le `while` sauf si ce dernier a été interrompu par un `break`.

````python
a = 0  # initialisation de la variable a
while a < 5:
    print(a)
    a += 1  # raccourci pour a = a + 1
````

````python
a = 0
while True:
    a += 1
    if a == 5:
        break
    print(a)
````

## 1.4 `match`

Cette instruction est relativement récente. Elle a été ajoutée à la version Python 3.10.

L'instruction `match` permet d'effectuer une action différente en fonction d'une valeur. Elle se combine avec plusieurs instructions `case` indentées et déclinant les différents cas. Toutes ces instructions nécessitent le caractère `:` en fin de ligne.

Le cas par défaut est matérialisé par le caractère `_`. Il est possible de joindre différents cas en utilisant le caractère `|`.

Il existe d'autres possibilités plus puissantes encore.

````python
# exemple d'usage de match
for i in range(10):
    
    print(i, end=" ")
    
    match i:
        case 0:
            print("Zéro")
        case 1:
            print("Un")
        case 2:
            print("Deux")
        case 4:
            print("Quatre")
        case 5:
            print("Cinq")
        case 3 | 6 | 9:              # plusieurs cas
            print("Multiple de 3")
        case _:                      # autres cas
            print("Autre valeur")
````

## 1.5 `try`

L'instruction `try` permet gérer les <u>exceptions</u>, c'est-à-dire d'éventuelles erreurs générées pendant l'exécution. Elle est suivie par une ou plusieurs instructions `except` qui gèrent les cas d'erreur. Toutes ces instructions nécessitent le caractère `:` en fin de ligne.

Il est possible d'intercepter les différents types d'erreurs :
- `NameError`, `ValueError`, `TypeError`, `ZeroDivisionError`, etc.
- avec la syntaxe : `except ZeroDivisionError:`
- on peut également accéder à l'objet Python qui matérialise l'erreur: `except ZeroDivisionError as err:`
- il est possible d'ajouter l'instruction `finally` qui s'exécute systématiquement
- il est possible d'ajouter l'instruction `else` qui ne s'exécute qu'en cas d'absence d'exception

Exemples simples (la gestion fine des erreurs est assez complexe) :

````python
# exemple simple de gestion d'erreur
try:
    1 / 0
except ZeroDivisionError:
    print("erreur : division par zéro")
````

````python
# exemple simple de gestion d'erreur
from math import sqrt

try:
    sqrt(-1)
except ZeroDivisionError:
    print("erreur : division par zéro")
except ValueError:
    print("erreur : racine d'un nombre négatif")
else:
    print("pas d'exception")
````

````python
# exemple simple de gestion d'erreur
try:
     1 + "a"
except TypeError as err:
    print("Erreur :", err)
finally:
    print("opération terminée")
````

# 2. Fonctions

## 2.1 Résumé

Qu'est-ce qu'une fonction en Python :
- **Blocs de code réutilisables** : Les fonctions sont des blocs de code réutilisables conçus pour accomplir une tâche spécifique, réduisant ainsi la redondance et améliorant l'organisation du code.
- **Définition et appel** : Elles sont définies en utilisant le mot-clé `def` suivi du nom de la fonction et de parenthèses, et elles sont appelées en utilisant le nom de la fonction avec des parenthèses.
- **Paramètres et arguments** : Les fonctions peuvent accepter des paramètres, ce qui leur permet de recevoir des valeurs en entrée (arguments) et de les manipuler. Il existe deux types d'arguments : les arguments positionnels et les arguments à mot-clés permettant de définir des valeurs par défaut.
- **Valeurs de retour** : Les fonctions peuvent retourner des valeurs en utilisant l'instruction `return`, ce qui permet d'utiliser le résultat de leur calcul ailleurs dans le programme.
- **Portée** : Les fonctions ont leur propre portée locale, ce qui signifie que les variables définies à l'intérieur d'une fonction ne sont pas accessibles en dehors de celle-ci, sauf si elles sont retournées ou déclarées globales.
- **Fonctions d'ordre supérieur** : En Python, les fonctions peuvent être passées en tant qu'arguments à d'autres fonctions, retournées comme valeurs et assignées à des variables, ce qui adhère aux paradigmes de la programmation fonctionnelle.

**Modèle d'une fonction**

```python
# définition de la fonction
def ma_fonction(x, y, z=1):
    """Cette fonction retourne la somme de x et y multipliée par z.
       Par défaut, z vaut 1."""
    
    resultat = (x + y) * z
    
    return resultat

# appels de la fonction
ma_fonction(1, 2, 3) => 9
ma_fonction(1, 2) => 3
```

## 1.4 Nombre indéfini d'arguments

On peut avoir un nombre indéfini d’arguments positionnels en utilisant `*args` (ils sont rassemblés dans un tuple) et aussi des arguments à mots-clef en utilisant `**kwargs` (ils sont rassemblés dans un dictionnaire).

````python
def test(a, b, *args):
    """Calcule a * b + somme(args)"""
    print(a, b, args)
    val = 0
    for i in args:
        val += i
    return a * b + val
````

````python
# appel de la fonction test
# le paramètre "a" est affecté avec le premier argument passé
# le paramètre "b" est affecté avec le deuxième argument passé
# le paramètre "args" est affecté avec le reste des arguments passés
test(5, 4, 3, 2, 1)
````

#### fonctions `map()` et `filter()`

Il existe 2 fonctions prédéfinies qui permettent d'appliquer des fonctions sur des séquences :
- `map` : `map(fonction, séquence)` renvoie un itérateur sur l'application de la fonction sur chaque élément de la séquence.
- `filter` : `filter(fonction, séquence)` renvoie un itérateur sur chaque élément de la séquence pour lesquels la fonction retourne `True`.

Pour visualiser le résultat, il faut le transformer ensuite en liste.

````python
# map
liste = list(map(len, ["A", "AB", "ABC"]))

print(liste)
````

````python
# filter
list(filter(lambda x: x.isalpha(), ["A10", "AB0", "ABC"]))
````

````python
# filter plus simple avec la méthode str.isalpha
list(filter(str.isalpha, ["A10", "AB0", "ABC"]))
````

#### Formatage des chaînes de caractères

Différentes techniques permettent de formater les chaînes de caractères :
- i) Les chaînes de caractères formatées (`f-strings`)
- ii) La méthode de chaîne de caractères `format()`
- iii) L'ancien opérateur `%`
- iv) La concaténation de chaînes avec l'opérateur `+`

##### i) f-string

A partir de Python 3.7, avec les `f-string`, les identifiants et les instructions sont placées entre `{}` directement dans la chaine de caractères.

````python
# f-string
prenom = "Marie"
age = 4
f"{prenom} a {age} ans"
````

````python
# f-string
a, b = 3, 4
f"{a} x {b} = {a * b}"
````

On peut utiliser des formats particuliers avec la syntaxe `{variable:format}` et on peut même faire apparaître le nom de la variable  avec la syntaxe `{variable=}` (pratique pour débugger avec Python 3.8+).

````python
# formattage des flottants
pi = 3.141592653589793
f"{pi:.4f}"
````

````python
# nom de la variable
name = "Daffy Duck"
hello = f"Hello! Mon nom est {name=}"
print(hello)
````

##### ii) méthode .format()

En Python 3, avec la méthode `.format()` les identifiants sont dans l'appel de méthode soit par mots-clés, soit en positionnel.

````python
# formattage des chaines avec la méthode format (mots-clés)
"{prenom} a bien {age} ans".format(prenom="Marie", age=4)
````

````python
# formattage des chaines avec la méthode format (positionnel)
"{1} a toujours {0} ans".format(4, "Marie")
````

````python
# formattage des chaines avec la méthode format (positionnel implicite)
"{} a bien {} ans".format("Marie", 4)
````

##### iii) ancienne méthode

Historiquement, en Python 2 et 3, avec l'opérateur `%` les identifiants sont dans le tuple passé à l'opérateur.

````python
# formattage des chaines avec l'opérateur %
"%s a %i ans" % ("Marie", 4)

#NB : s pour string, d ou i pour entier signé, f pour un décimal
````

##### iv) la concaténation de chaine (basique et à éviter)

````python
# formattage des chaines avec l'opérateur +
prenom + " a bien " + str(age) + " ans"
````

## 3.2 Fichiers

#### Fichiers textes ou binaires

La fonction `open()` permet d'ouvrir des fichiers textuels et binaires en mode lecture, écriture ou ajout. Les attributs correspondant au type de fichier et au mode d'ouverture sont à indiquer dans une chaîne de caractères passée en paramètre (par défaut le paramètre est `"rt"` (lecture seule en mode texte).

Les fichiers sont en encodage Unicode (UTF-8) par défaut en Python 3. Il est possible de préciser un autre encodage, par exemple `encoding="latin-1"`, à l'appel de la fonction.

Caractère|Signification
-|-
*r*|lecture (par défaut)
*w*|création (avec écrasement si le fichier existe) et écriture
*x*|création exclusive (avec erreur si le fichier existe) et écriture
*a*|création (si le fichier n'existe pas) et écriture à la fin d'un fichier existant
*b*|mode binaire (vs texte)
*t*|mode texte (défaut)
*+*|lecture et écriture

Cette fonction retourne un objet de type `file` dont la classe exacte dépend de la nature du fichier et du mode d'ouverture. Une fois utilisé le fichier doit être refermé avec la méthode `close()`.

Il est possible et conseillé d'ouvrir un fichier avec le gestionnaire de contexte `with`. Une fois utilisée la ressource sera fermée automatiquement.

Cet objet peut répondre à des méthodes de lecture et d'écriture.

Méthode|Signification
-|-
*read()*|lit le contenu entier du fichier
*readline()*|lit une ligne du fichier
*write()*|écrit dans le fichier
*seek()*|positionne le curseur du fichier
*tell()*|donne la position du curseur du fichier

Quelques exemples :

````python
# écriture d'un fichier

f = open("data/data.txt", "w")  # ouverture du fichier texte 'data.txt' en écriture
f.write("ceci est un fichier tout simple.")
f.close()
````

#### Le gestionnaire de contexte `with`

````python
# écriture d'un fichier avec un gestionnaire de contexte

with open("data/data.txt", "w") as f:  # idem avec un gestionnaire de contexte
    f.write("Ceci est un fichier tout simple.")
````

````python
# lecture d'un fichier

with open("data/data.txt") as f:  # lecture du fichier
    content = f.read()

# affichage de son contenu
print(content)
````

Cet objet peut également être considéré comme un itérateur pour une boucle `for`. Dans ce cas chaque élément correspond à la valeur retournée par `readline()` jusqu'à épuisement du fichier.

````python
with open("data/data.txt", "w") as f:
    f.write("Ceci est un fichier tout simple.\n")
    f.write("Deuxième ligne.\n")
    f.write("Troisième ligne.\n")

with open("data/data.txt") as f:
    for line in f:
        print(line)
````

#### Fichiers JSON

Pour sauvegarder des objets complexes comme des listes ou bien des dictionnaires, il est possible d'utiliser le format *JSON* (JavaScript Object Notation) qui présente l'avantage d'être compris par d'autres langages.

L'opération qui consiste à transformer un objet informatique complexe en un texte linéraire s'appelle la *sérialisation*.

````python
# objet Python structuré
clients = [{"nom": "Dupuis", "prénom": "Xavier"},
           {"nom": "Dupont", "prénom": "Emma"},
           {"nom": "Martin", "prénom": "Pierre"} ]
````

````python
import json

# on sauvegarde le dictionnaire avec la librairie json
with open("clients.json", "w") as f:
    json.dump(clients, f)
    
# on relit le dictionnaire sauvegardé
with open("clients.json") as f:
    clients2 = json.load(f)

print(clients2)
````

````python
# on accède bien à l'objet structuré
clients2[0]["nom"]
````

## 3.3 Bases de données

Python est capable de s'interconnecter avec la plupart des bases de données du marché qui implémentent le standard *ODBC* (Open Database Connectivity.

#### sqlite3

Le module `sqlite3` gère directement la connexion avec les bases de données *sqlites. La fonction `connect()` crée une connexion avec une base, la méthode `.cursor()` permet d'obtenir un curseur sur la base qui effectue les requêtes avec `.execute()` et retourne chacun des résultats avec `.fetchone()` à utiliser de préférence plutôt qu'un `.fetchall()`.

````python
# exemple avec une micro-base Sqlite3
# on alimente une table

import sqlite3

with sqlite3.connect("data/salaries.sqlite") as db:
    cursor = db.cursor()
    
    # show table columns
    print("Table Salaries")
    print("="*40)
    cursor.execute("""PRAGMA table_info(Salaries);""")
    # boucle pour obtenir les résultats
    row = cursor.fetchone()
    while row:
        print(row)
        row = cursor.fetchone()
    print()
    
   # sélection des enregistrements d'une table
    print("Employees with TotalPay > 400000")
    print("="*40)
    cursor.execute("""SELECT Id, EmployeeName, JobTitle, TotalPay FROM Salaries WHERE TotalPay > ?;""", (400000,))  # passage de paramètres pour éviter les injections SQL
    # boucle pour obtenir les résultats
    row = cursor.fetchone()  # à préférer à fetchall() qui retourne TOUS les résultats
    while row:
        print(*row)
        row = cursor.fetchone()
    print()
````
