# 1. Introduction

Un script Python est un fichier d'extension `.py` qu'il est possible de soumettre à un interpréteur Python pour qu'il soit exécuté.

Les modules constituent des extensions du langage Python. Un module Python est composé d'un fichier d'extension *.py* définissant des objets, des fonctions ou des classes.

Un package regroupe plusieurs modules dans différents sous-répertoires.

Il peut s'agir des modules essentiels livrés avec le langage (par exemple, **sys**, **os**, **math**, **re**, **urllib**, ...), de modules tiers (par exemple, **NumPy**, **SciPy**, **matplotlib**, **pandas**, **scikit-learn**, **polars**, ...) et aussi des modules créés par vous-même.

On peut utiliser un module ou un package avec l'instruction `import`. On utilise soit le nom du fichier (sans l'extension) soit le chemin qui qui mène au fichier (séparé par des points).

Par défaut l'espace de nom est celui du module ou du package, c'est-à-dire, pour être reconnus les objets définis par le module (fonctions, classes) doivent être préfixés par le nom du module. Il est possible de changer le nom ou de s'en passer mais attention aux collisions.

Lors de la première utilisation d'un module ou du package, celui-ci est "précompilé" dans un fichier d'extension `.pyc` qui sera utilisé pour les appels suivants. Ces fichiers contiennent les *bytes codes* qui sont exécutés par la machine virtuelle Python. Avec IPython Notebook les fichiers "précompilés" sont sauvegardés dans le répertoire `__pycache__`.

## 2.6 module **collections**

Ce module définit des types supplémentaires pour contenir des objets

#### nametuple()

Cette fonction permet de créer un tuple nommé qui ressemble à une instance de classe sans méthode associée.

Un *namedtuple* permet d'accéder à ses différentes valeurs en utilisant un attribut plutôt qu'une position.

````python
# points sans namedtuple

p = (1, 2)
print(p)
print(p[0]) # x
print(p[1]) # y
````

````python
# points avec namedtuple
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
p = Point(1, 2)
print(p)
print(p.x) # x
print(p.y) # y
````

#### defaultdict

Cette sous-classe de `dict` comprend un constructeur d'objets par défaut qui est utilisé lorsqu'une clé est absente.

````python
# compter les mots d'une phrase sans defaultdict

phrase = "a bon chat bon rat"
dico = {}

for mot in phrase.split():
    if mot in dico:
        dico[mot] += 1
    else:
        dico[mot] = 1

print(dico)
````

````python
# defaultdict
from collections import defaultdict

dico = defaultdict(int)  # int représente ici la fonction int() qui retourne 0

for mot in phrase.split():
    dico[mot] += 1

print(dico)
````

#### Counter

Cette sous-classe de `dict` compte automatiquement les objets hashables.

````python
# Counter
from collections import Counter

c = Counter()

c.update(phrase.split())

print(c)
````

## 2.8 module **re**

Le module **re** fournit des opérations sur les expressions régulières similaires à celles que l'on trouve dans le langage Perl.

Les expressions régulières (regex ou expressions rationnelles) sont formées d'un petit langage extrêmement puissant, spécialisé et disponible en Python avec le package **re**.

En utilisant ce langage, on peut spécifier des motifs (patterns) pour l'ensemble des chaînes que l'on souhaite filtrer ou capturer.
- `.`: matche n'importe quel caractère
- `*`: matche 0 ou plusieurs répétitions du motif précédent
- `+`: matche 1 ou plusieurs répétitions du motif précédent
- `?`: matche 0 ou 1 répétition du motif précédent
- `^`: matche le début de la chaîne
- `$`: matche la fin de la chaîne
- `|` : alternative en 2 motifs
- `\`: si l'on veut utiliser un caractère spécial des regex, il faut le préfixer par un `\`

Il est également possible de lister entre crochets les caractères que l'on souhaite matcher:
- `[aeiouy]`: matche n'importe quelle voyelle
- `[^aeiouy]`: ne matche aucune voyelle (ici le `^` est interprété comme un NON)
- `[A-Z]`: matche tous les caractères compris entre `A` et `Z` (plusieurs paires ordonnées peuvent être utilisées: e.g., pour les majuscules, minuscules, lettres accentuées et les chiffres : `[A-Za-zÀ-ÿ0-9]`)

Il existe heureusement des raccourcis pour nommer toutes les lettres, tous les chiffres, les ponctuations... :

Elément| Description | Equivalent
-|-|-
`.`	| Tout sauf `\n`| `[^\n]`
`\d`	| Tous les chiffres | `[0-9]`
`\D`	| Tout sauf les chiffres | `[^0-9]`
`\s`	| Les espaces et équivalents | `[ \t\n\r\f\v]`
`\S`	| Tout sauf espaces et équivalents | `[^ \t\n\r\f\v]`
`\w`	| Tous les caractères alphanumériques | `[a-zA-Z0-9_]`
`\W`	| Tout sauf les caractères alphanumériques | `[^a-zA-Z0-9_]`

Le langage comporte d'autres fonctionalités puissantes:
- raccourcis : cf. ci-dessus
- répétitions : `*`, `+`, `?`, `{m,n}`
- motifs de capture : `(...)`
- motifs de non capture : `(?:...)`
- référence à un groupe nommé : `(P=name...)`
- assertions en avant : `(?=...)`
- anti-assertions en avant : `(?!...)`
- anti-assertions en arrière : `(?<=...)`

Différentes fonctions sont disponibles :

- `match()` retourne un objet de type `Match` si la chaîne débute par le motif.
- `search()` retourne un objet de type `Match` si la chaîne contient le motif.
- `findall()` permet d'extraire toutes les chaînes qui matchent avec un motif.
- `finditer()` permet d'itérer sur tous les objets de type `Match` correspondant à un motif.
- `sub()` substitution d'un motif.

Il est également possible d'instancier une regex avec la fonction `compile()` et d'utiliser ensuite les fonctions ci-dessus comme des méthodes.

Il existe également des *flags* qui permettent de modifier certaines interprétations globales : `re.I` ignore case, `re.M` multi-lignes, etc.

Pour éviter les problèmes avec le caractère `\`, il faut préfixer les chaînes des motifs de recherche par un `r` (raw).

````python
# exemple

import re

texte = """Python is a programming language that lets you work quickly and integrate systems more effectively."""

print(re.findall(r"(\w+)", texte))
print(re.findall(r"(\w*o\w*)", texte))
````

## 2.8 module **datetime**

Le module **datetime** permet de manipuler des données temporelles en Python.

## 2.9 module **unitest**

Les tests unitaires permettent de vérifier automatiquement qu'une fonction se comporte comme prévu, sans avoir à relancer manuellement le programme à chaque modification. En Python, le module standard **unittest** fournit un cadre pour écrire ces tests sous forme de classes héritant de `unittest.TestCase`, avec des méthodes dont le nom commence par "test_". Chaque méthode contient des assertions (`assertEqual`, `assertTrue`, `assertRaises`, etc.) qui comparent le résultat obtenu au résultat attendu. C'est une pratique essentielle en data science pour sécuriser les fonctions de nettoyage ou de transformation de données, qui sont souvent modifiées au fil d'un projet.

Le module plus complet **pytest** est également très utilisé.

````python
import unittest
import statistics

def nettoyer_donnees(donnees):
    return [d for d in donnees if d.get("age") is not None]

def calculer_moyenne_age(donnees):
    ages = [d["age"] for d in donnees]
    return statistics.mean(ages)

class TestNettoyage(unittest.TestCase):

    def setUp(self):
        self.donnees = [
            {"nom": "Alice", "age": 25},
            {"nom": "Bob", "age": None},
            {"nom": "Carla", "age": 30},
        ]

    def test_suppression_valeurs_manquantes(self):
        resultat = nettoyer_donnees(self.donnees)
        self.assertEqual(len(resultat), 2)

    def test_moyenne_age(self):
        resultat = nettoyer_donnees(self.donnees)
        moyenne = calculer_moyenne_age(resultat)
        self.assertAlmostEqual(moyenne, 27.5)

unittest.main(argv=[''], verbosity=2, exit=False)  # pour le notebook
# unittest.main()  # en dehors d'un notebook
````

# 3. Modules, Packages et Scripts

## 3.1 Création d'un module


Il est très simple de faire appel à des fonctions ou à des classes depuis un autre fichier.

Il suffit de stocker vos classes et fonctions dans un fichier d'extension `.py`. Celui-ci pourra être stocké dans un répertoire appartenant aux chemins de Python ou dans votre répertoire de travail.

Vous avez alors la possibilité de faire :
```python
import monscript
```
et d’utiliser les fonctions et classes de ce fichier.

Vous venez de créer un module.

**Attention ceci n’est pas un package !**

Il est possible de définir ses propres modules. Par exemple, on sauvegarde ce code dans un fichier `hyp.py`.

```python
# coding: utf-8

from math import sqrt

def hypotenuse(x, y):
    return sqrt((x * x) + (y * y))
```

Ensuite on peut utiliser cette fonction dans un programme.

````python
# import de la fonction hypothenuse du module hyp
from hyp import hypotenuse

hypotenuse(3, 4) + hypotenuse(5, 12)
````

## 3.2 Création et exécution d'un script

Pour exécuter un script Python enregistré dans un fichier `monscript.py` par exemple, il suffit d'ouvrir un terminal et de lancer la commande :
<pre>
python monscript.py
</pre>

Il existe une syntaxe précise à placer dans le script pour exécuter un code spécifique lorsqu'un fichier est exécuté comme un script et non importé comme module :

```python
if __name__ == "__main__":
    # execute only if run as a script
    main()
```

La variable `__name__` vaut `"__main__"` si le script est exécuté avec l'instruction `python monscript.py`, mais elle vaut `"monscript"` si le script est importé.

Dans le cas ci-dessus, si le script est exécuté avec la commande `python monscript.py` la fonction `main()` sera automatiquement exécutée.

SInon, il faut l'exécuter explicitement, par exemple, `python -c "from monscript import main; main()"`.

## 3.3 Création d'un package

Un package est constitué de plus d’informations qu’un simple fichier d'extension `.py`.

Voici les étapes pour créer votre premier package :
- Choisissez un nom simple `mon_package` et traduisant bien ce que va faire votre package (si possible un nom qui n’est pas déjà utilisé)
- Créez un répertoire du nom de votre package : `mon_package`
- Créez un fichier `__init__.py` dans ce répertoire, ce fichier peut être vide dans un premier temps
- Dans le même répertoire ajoutez des fichiers `mon_fichier.py` avec vos fonctions et vos classes

Pour importer votre package dans votre notebook ou dans un programme, il vous suffit d'utiliser :

```python
import mon_module.mon_fichier
```

Si votre objectif est de publier le package sur PyPI, il faudra ajouter un fichier setup et les dépendances associées.

#### Quelques détails sur la création de package

Le fichier `__init__.py` peut être vide mais il peut aussi contenir des informations.

Ce fichier est lancé à chaque chargement du package, il peut inclure des dépendances, des vérifications de versions, des imports vers les fonctions et classes des fichiers de votre package...

Où peut-on mettre un package ?
- Par défaut, si le package se trouve dans le même répertoire que le script exécuté, ça fonctionne
- Sinon, le package doit se trouver dans un répertoire lié au `PYTHONPATH`
- Pour rechercher les répertoires Python existants, on utilise la variable `sys.path`
- Pour ajouter un répertoire "temporaire", on utilise la fonction `sys.path.insert()`
