# 1. Programmation par objets

## 1.1 Résumé

Qu'est-ce qu'une classe en Python :
- **Modèle pour les objets** : Une classe en Python est un modèle pour créer des objets (instances), encapsulant des données et le comportement associé à ces données.
- **Attributs et méthodes** : Elle contient des attributs (variables d'instance) et des méthodes (fonctions) qui définissent les propriétés et comportements des objets créés à partir de la classe.
- **Encapsulation** : Les classes permettent l'encapsulation, en regroupant les données et les fonctions ensemble.
- **Héritage** : Les classes peuvent hériter des attributs et des méthodes d'autres classes, ce qui permet la réutilisation du code et la création de structures de classes hiérarchiques.
- **Instantiation** : Les objets sont créés à partir de classes via l'instanciation, qui consiste à appeler la classe comme s'il s'agissait d'une fonction.
- **Invocation de méthode** : Les méthodes d'une classe sont invoquées en utilisant la notation par point sur une instance de la classe, permettant ainsi aux objets d'exécuter les actions définies dans la classe.

## Modèle d'une classe

```python
# définition de la classe
class Personne():
    """Cette classe représente une personne avec un nom et un âge.
    Une personne est capable de dire bonjour à un nom donné en mentionnant son propre nom et son âge."""
    
    def __init__(self, nom, age):
        """Méthode d'initialisation.
        Ici, nous stockons dans la variable d'instance "nom" la chaîne de caractères passée lors de la création d'une instance,
```

## 1.3 Classes définies par l'utilisateur

#### Exemple avec la classe Point

La classe Point définit un point en 2 dimensions.

````python
class Point:
    
    def __init__(self, a=0.0, b=0.0):
        """
        Un point est défini par ses deux coordonnées (x,y)
        Le point par défaut est l'origine (0, 0)
        """
        self.x = a
        self.y = b

    
    def __str__(self):
        """
        Méthode implicite utilisée par print()
        Quand on print un point, on veut voir ses coordonnées
        """
        return(f'({self.x},{self.y})')
    
    def __eq__(self, p):
        """
        Méthode implicite utilisée par ==
        Deux points sont égaux ssi ils on les mêmes coordonnées
        """
        return (self.x == p.x) and (self.y == p.y)

    # METHODES
    
    def est_diagonal(self):
        """
        Méthode qui s'applique au point seul
        Un point est sur la diagonale si x = y
        """
        return self.x == self.y

    def transpose(self):
        """
        Méthode qui s'applique au point seul
        On crée un nouveau point en échangeant ses coordonnées
        """
        return Point(self.y, self.x)

    def milieuAvec(self, autrePoint):
        """"
        Méthode qui nécessite un deuxième point en argument pour trouver le milieu
        On crée un nouveau point, au milieu des 2 autres
        """
        return Point((self.x + autrePoint.x) / 2, (self.y + autrePoint.y) / 2)
````

Pour créer une instance, on invoque sa classe comme une fonction `Point()` avec les paramètres nécessaires.

````python
# Initialisation d'un point par défaut
p1 = Point()
print(p1)
````

#### Exemple avec la classe Polygone

Soit une nouvelle classe Polygone qui va ré-utiliser la classe Point.

Un polygone est composé d'une liste d'instances de la classe Point. Il possède des méthodes permettant de lui ajouter un ou plusieurs points ainsi qu'un une autre instance de la classe Polygone.

````python
class Polygone:
    def __init__(self):
        """
        Un polygone est fait d'une liste de points
        """
        self.points = []

    def __str__(self):
        """
        Méthode implicite utilisée par print()
        Quand on print un polygone, on veut voir sa liste de points
        """
        return("Poly{%s}" % ', '.join([str(p) for p in self.points]))

    def ajout_point(self, point):
        """
        Ajout d'un point au polygone
        On s'assure qu'il n'y a pas de doublon
        """
        if not point in self.points: # garanti que tous les points sont différents
            self.points.append(point)

    def ajout_points(self, points):
        """
        Ajout de plusieurs points au polygone
        Appels successifs de la méthode ajout_point()
        """
        for point in points:
            self.ajout_point(point)

    def ajout_polygone(self, polygone):
        """
        Ajout des points d'un autre polygone au polygone
        On passe par la méthode ajout_points()
        """
        self.ajout_points(polygone.points)

    def nom(self):
        """
        Détermine le nom du polygone en fonction du nombre de points
        """
        nb = len(self.points)
        
        if nb < 6:
            noms = ["vide", "point", "segment", "triangle", "quadrilatère", "pentagone"]
            
            return noms[nb]
            
        return "polygone"
````

````python
p = Polygone()
print(p, p.nom())
````

````python
p.ajout_point(p1)
p.ajout_point(p2)
print(p, p.nom())
````

````python
q = Polygone()
q.ajout_points([p2, p3, p4])
print(q, q.nom())
````

````python
p.ajout_polygone(q)
print(p, p.nom())
````

## 1.4 Méthodes statiques et méthodes de classes

Il existe des décorateurs qui permettent de définir des méthodes statiques et des méthodes de classe.

- `@staticmethod` : méthode statique, la méthode ne fait pas appel aux variables d'instance, pas de variable `self`.
- `@classmethod` : méthode de classe, variable `cls` à la place de la variable `self`.

````python
class CompteBancaire:
    
    def __init__(self, numero, valeur=0):
        self.numero = numero
        self.valeur = valeur

    def __str__(self):
        return f"[n°:{self.numero}|valeur={self.valeur}]"

    def ajout(self, montant):
        self.valeur += montant

    @staticmethod
    def message(texte):
        print(texte)

    @classmethod
    def whoami(cls):
        print(f"Je suis un {cls.__name__}")
````

````python
a = CompteBancaire("1234")
print(a)
a.ajout(100)
print(a)
a.message("hello")
a.whoami()
````

<div class="alert alert-info">
    <h3><i class="fa fa-info-circle"></i>  Astuces</h3>
    <ul>
        <li>L'invocation de la méthode par la notation <code>a.ajout(100)</code> (sucre syntaxique) est équivalent à l'appel fonctionnel <code>CompteBancaire.ajout(a, 100)</code>, avec passage de l'instance en premier argument, d'où la variable <code>self</code> dans les méthodes.</li>
        <li>L'invocation de la méthode statique par la notation <code>a.message("hello")</code> est équivalent à l'appel fonctionnel <code>CompteBancaire.message("hello")</code>, pas de passage de l'instance en argument.</li>
        <li>L'invocation de la méthode de classe par la notation <code>a.whoami()</code> est équivalent à l'appel fonctionnel <code>CompteBancaire.whoami(CompteBancaire)</code>, passage de la classe au lien de l'instance en premier argument.</li>
    </ul>
</div>

## 1.5 Héritage des classes

Il existe un mécanisme d'héritage simple (et multiple, non abordé).

La fonction `super()` permet d'invoquer une méthode de la superclasse qui est utilisée notamment lors de l'initialisation des instances.

````python
class CompteDevise(CompteBancaire):
    
    def __init__(self, numero, valeur=0, devise="EUR"):
        super().__init__(numero, valeur)
        self.devise = devise

    def __str__(self):
        return f"[n°:{self.numero}|valeur={self.valeur} {self.devise}]"
````

````python
b = CompteDevise("567890", valeur=100, devise="CHF")
print(b)
b.ajout(1000)
print(b)
b.message("goodbye")
b.whoami()
````

#### création d'un générateur avec une *list comprehension*

La notation avec des `[...]` qui crée des *list comprehension* peut être transformée en générateur en remplaçant les crochets par des parenthèses.

Les 2 peuvent ensuite être utilisés dans une boucle par exemple.

````python
# sans générateur

l = [i * i for i in range(1_000_000)]
print("les 10 premiers :", l[:10])
print("les 10 derniers :", l[-10:])

import sys

print("empreinte mémoire :", sys.getsizeof(l))
````

````python
# avec générateur

g = (i * i for i in range(1_000_000)) # notation list comprehension avec des () à la place des []
print("les 3 premiers :", next(g), next(g), next(g))

import sys

print("empreinte mémoire :", sys.getsizeof(g))
````

#### instruction `yield`

Dans une fonction, l'instruction `yield` dans une boucle permet de transformer la fonction en un générateur.

La fonction est exécutée jusqu'à l'instruction `yield` et retourne une valeur. Elle reprend ensuite à partir de là où elle s'était interrompue, etc.

````python
# générateur avec yield

def compteur(maximum):
    n = 0
    while n < maximum:
        yield n
        n += 1
````

````python
# utilisation du générateur

c = compteur(10)

print("première valeur :", next(c))
print("deuxième valeur :", next(c))

while True:
    try:
        print("enième valeur :", next(c))
    except StopIteration:
        break
````

````python
c = compteur(20)

for x in c:
    print(x)
````

## 2.2 Décorateurs

En Python, les décorateurs sont des fonctions qui s'appliquent à d'autres fonctions en permettant d'enrichir leur comportement <u>sans modifier le code initial</u>.

#### Un exemple simple

````python
# exemple simple

def bonjour():
    print("Bonjour !")

bonjour()
````

````python
# fonction s'appliquant à une fonction

def executer_fonction(func):
    print("Je vais exécuter la fonction passée en argument...")
    func()
    print("... la fonction passée en argument a été exécutée.")
````

````python
# on exécute la fonction executer_fonction() appliquée à la fonction bonjour()
executer_fonction(bonjour)
````

````python
@executer_fonction
def bonsoir():
    print("Bonsoir !")
````

#### avec un décorateur

L'utilisation d'un décorateur s'effectue en plusieurs temps :
- On définit un décorateur
- On l'applique à la définition d'une fonction
- On exécute la fonction décorée

````python
# 1. création du décorateur
def mon_decorateur(func):
    def wrapper(*args, **kwarg):
        print("Avant la fonction")
        value = func(*args, **kwarg)
        print("Après la fonction")
        return value
    return wrapper
````

````python
# création de la fonction avec le décorateur
@mon_decorateur
def carre(x):
    return x * x
````

````python
# exécution de la fonction décorée
print(carre(2026))
````

#### Librairie **functools**

La librairie **functools** comporte de nombreux utilitaires pour des fonctions.

Par exemple, le décorateur `@cache` qui permet de mémoriser les valeurs déjà calculées par la fonction.

````python
# sans cache
def fibonacci(n):
    if n < 2:
        return 1
        
    return fibonacci(n-1) + fibonacci(n-2)
````

````python
%%time
fibonacci(36)
````

````python
# avec cache
from functools import cache

@cache
def fibonacci(n):
    if n < 2:
        return 1
        
    return fibonacci(n-1) + fibonacci(n-2)
````

````python
%%time
fibonacci(36)
````

#### Ecrire un wrapper avec functools.wraps

````python
# sans functools.wraps

import time

def timer1(func):

    def wrapper(*args, **kwargs):

        debut = time.perf_counter()

        resultat = func(*args, **kwargs)

        fin = time.perf_counter()

        print(
            f"{func.__name__} : "
            f"{fin - debut:.6f} secondes"
        )

        return resultat

    return wrapper
````

````python
# utilisation du wrapper timer1

@timer1
def add(x, y):
    time.sleep(1)
    return x + y
````

````python
# utilisation de la fonction

add(1, 2)
````

````python
# la fonction a perdu son nom !

add.__name__
````

````python
# avec functools.wraps

from functools import wraps

def timer2(func):

    @wraps(func)
    def wrapper(*args, **kwargs):

        debut = time.perf_counter()

        resultat = func(*args, **kwargs)

        fin = time.perf_counter()

        print(f"{func.__name__} : {fin - debut:.6f}s")

        return resultat

    return wrapper
````

````python
# utilisation du wrapper timer2

@timer2
def add(x, y):
    time.sleep(1)
    return x + y
````

````python
# utilisation de la fonction

add(1, 2)
````

````python
# la fonction a encore son nom !

add.__name__
````

## 2.3 Typage des fonctions

Python est un langage dynamiquement typé : il n'est pas nécessaire de déclarer explicitement le type des variables. Il est néanmoins possible d'indiquer les types attendus par une fonction grâce aux annotations de type (*type hints*). Elles permettent de rendre le code plus lisible, facilitent la compréhension et peuvent être exploitées par les IDE et des outils d'analyse statique.

Les types des paramètres sont indiqués après leur nom, à l'aide de `:`, et le type de retour est indiqué après `->` :

```python
def addition(a: int, b: int) -> int:
    return a + b

addition("hello", "world")
```

Les annotations ne forcent pas Python à vérifier les types lors de l'exécution :

Ce code peut tout de même s'exécuter et retourner `"helloworld"`. Les annotations constituent donc avant tout une information destinée au programmeur et aux outils, et non une contrainte du langage.

On peut également annoter des fonctions avec des valeurs par défaut (l'annotation se place avant la valeur par défaut) et qui ne retournent aucune valeur avec `None` :

```python
def afficher_resultat(resultat: float = 0) -> None:
    print(resultat)
```

Les annotations peuvent utiliser les types Python courants :

```python
def f(
    age: int,
    nom: str,
    score: float,
    actif: bool
) -> str:
    ...
```

Elles peuvent aussi décrire des structures de données :

```python
def compter(mots: list[str]) -> dict[str, int]:
    ...
```

Enfin, les annotations peuvent être utilisées pour documenter plus précisément les fonctions complexes, notamment avec `Optional`, `Union`, `Callable` ou les types génériques du module `typing`.
